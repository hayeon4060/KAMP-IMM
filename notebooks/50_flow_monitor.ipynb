| 신호 (라벨 미사용) | 내용 | 나온 곳 |
|---|---|---|
| S1 EWMA·std4 | 흔들림 점수의 지수 이동평균 | 46번 RG3 미성형 (사후 선택) |
| S2 과열 드리프트 | 배럴2~5·호퍼 최근 30샷 평균 − 재가동 11~30샷 기준선 (**고정 기준선**) | 47번: 실제 가스 = 약한 과열(0.2~0.5배) → 누적 필요 |
| S3 계량 수준 이동 | 계량시간·사이클 최근 10샷 평균 − 고정 기준선 | 47·48번: RG3 미성형 = 계량지연형 |
| S4 변수 관계 변화 | 배럴·호퍼·계량시간 15쌍의 20샷 이동 상관이 정상과 다른 정도 | 새 가설 (히터 제어·공급 불안정) |

**46번에서 고친 점**: 재가동 후 10샷은 경보 제외(Run 시작 착시 제거), 누적형 대신 창(window) 방식(기준 보정 문제 회피)

**검증**
- `p_` = 원형 이동 순열 / **`보정p_` = 신호 6개 × 경보율 3개 중 최고를 고른 효과까지 뺀 p** ← 이게 0.05 아래여야 '신뢰'
- `[같은 K]` = 같은 검사 샷 수로 단발 std4 / 최종 모델 점수를 썼을 때
- 47번 주입 실험 재사용: 새 신호가 서서히 변하는 이상·계량지연을 잡는지
- 실행 시간: 본 실행 수 분 + 주입 확인 제품당 수 분 + (선택) RG3 결합 순열 수십 분

```python
import pandas as pd
pd.set_option('display.width', 250); pd.set_option('display.max_columns', 40); pd.set_option('display.max_rows', 300)
%load_ext autoreload
%autoreload 2
import flow_monitor as FM
R = FM.run(products=('cn7', 'rg3'), n_perm=1000, n_adj=2000)
```

```python
display(FM.view(R, 'cn7', 0.05).round(3))
display(FM.view(R, 'cn7', 0.10).round(3))
```

## RG3 — 경보율 5% / 10%

```python
display(FM.view(R, 'rg3', 0.05).round(3))
display(FM.view(R, 'rg3', 0.10).round(3))
```

## RG3 미성형 EWMA·std4 — 46번 전체 탐색(방법 17 × 경보율 4) 기준으로 보정한 p
`보정p_미성형`이 0.05보다 크면 46번에서 본 3/8·5/8은 '많이 찾아보면 나올 수 있는 수준'

## 47번 주입 실험으로 새 신호 확인 (정상 기준 경보율 1%) — 최종 모델이 못 잡던 패턴을 잡나

```python
SC = FM.synthetic_check(('cn7', 'rg3'), n_ep=150, R=R)
for p in ('cn7', 'rg3')
    print(f'===== {p} 에피소드 탐지율')
    display(FM.synthetic_pivot(SC, p))
    print(f'===== {p} 뒤 5샷 탐지율 (지속 이상을 끝까지 잡나)')
    display(FM.synthetic_pivot(SC, p, '뒤 5샷 탐지율'))
```

## RG3: 가스 RF + 라벨 없는 미성형 후보 결합 vs N1 (시드 3개 평균)

## 읽는 법
1. **보정p < 0.05** 인 신호만 '흐름으로 탐지된다'고 주장 가능. p_만 낮고 보정p가 높으면 '찾아보면 나오는 수준'
2. `불량 없는 Run 경보율`이 목표 경보율과 비슷해야 현장 기준으로 그대로 쓸 수 있음 (46번 CUSUM 문제 재발 여부)
3. 주입 실험: S2가 가스_과열·미성형_냉각점도를, S3가 가스_계량지연을 **뒤 5샷까지** 잡으면 → 고정 기준선이 47번 한계(적응)를 해결
4. 주입에서는 잡는데 실제 불량에서 못 잡으면 → '실제 불량은 그 패턴/크기가 아니다' (데이터 한계로 보고)
5. RG3 결합: N1(라벨 사용)과 비슷하면 **라벨 없이도 같은 성능** → 현장 적용성 측면 장점 (단, 순열 p 확인)

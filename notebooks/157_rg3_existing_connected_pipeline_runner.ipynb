미성형 비라벨 적용은 RUN_UNLABELED_APPLICATION=True로145번을 추가 실행할 수 있다.

```python
from pathlib import Path
import nbformat,pandas as pd
from nbclient import NotebookClient
root=Path.cwd()
if not (root/'reports').exists():root=root.parent
# Existing model code only: no new models or thresholds invented in this runner.
steps=['133_rg3_shortshot_complementary_nested_ensemble.ipynb','139_rg3_shortshot_fixed_seed_stability.ipynb','141_rg3_shortshot_rare_joint_confirmation.ipynb','143_rg3_shortshot_fixed_candidate_review.ipynb','152_rg3_gas_then_shared_shortshot_metrics.ipynb','153_rg3_gas_then_best_shortshot_metrics.ipynb']
RUN_UNLABELED_APPLICATION=False
if RUN_UNLABELED_APPLICATION:steps.append('145_rg3_shortshot_unlabeled_application.ipynb')
for name in steps:
 print('Running existing notebook',name,flush=True)
 path=root/'notebooks'/name
 notebook=nbformat.read(path,as_version=4)
 NotebookClient(notebook,timeout=900,kernel_name='python3',resources={'metadata':{'path':str(root)}}).execute()
 nbformat.write(notebook,path)
 print('Completed',name,flush=True)
metric=pd.read_csv(root/'reports/tables/rg3_153_gas_best_shortshot_cascade_metrics.csv')
for evaluation,counts in [('gas_only_583',[8,6,9,560]),('shortshot_best_alone_574',[6,11,2,555]),('final_any_defect_591',[14,17,11,549])]:
 assert metric.loc[metric.evaluation.eq(evaluation),['TP','FP','FN','TN']].values.tolist()==[counts],evaluation
print(metric.to_string(index=False))
print('Development predictions reproduced; this is not an unseen test or proof of new-data accuracy.')
```
